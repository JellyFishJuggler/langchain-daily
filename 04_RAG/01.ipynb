```python
with open("data/dummy.txt", "w") as file:
    file.write("This is some dummy content.\n")
    file.write("Jupyter notebook example file.")
```

```python
# from langchain.document_loader import TextLoader
from langchain_community.document_loaders import TextLoader

loader = TextLoader('data/dummy.txt',encoding='utf-8')
loader
```

```python
print(loader.load())
```

```python
# Directory Loader
from langchain_community.document_loaders import DirectoryLoader

dirLoader = DirectoryLoader(
    'data',
    glob='*.txt',
    loader_cls=TextLoader,
    loader_kwargs={"encoding":"utf-8"},
    show_progress=False
)
dirLoader
print(dirLoader.load())
```

```python
# Directory Loader
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

pdfLoader = DirectoryLoader(
    'data',
    glob='*.pdf',
    loader_cls=PyMuPDFLoader,
    show_progress=False
)
print(pdfLoader.load())
```
